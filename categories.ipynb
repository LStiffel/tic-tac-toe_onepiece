```python
import pandas as pd
```

```python
df = pd.read_json("characters.json")
```

```python
df.info()
```

```python
characters = df.drop(columns=["image","first_appearance_arc","description","journey","relationships","epithet","age","height","image_pre"])
```

```python
characters.head()
```

```python
devil_fruits = pd.read_json("devil_fruits.json")
```

```python
devil_fruits.head()
```

```python
len(devil_fruits[devil_fruits["name"]=="bat SMILE"])
```

```python
def findtype(fruit_name):
    if(type(fruit_name) is list):
        
        return(["Logia","Paramecia"])
        
    fruit = devil_fruits[devil_fruits["name"]==" "+fruit_name] #This line is to fix, for example "bat SMILE" has 2 spaces before
    fruit.reset_index(inplace=True)
    if(len(fruit)>0):
        return fruit["type"][0]
    else:
        return("Unknown")
```

```python
print(findtype("bat SMILE"))
```

```python
characters["devil_fruit"] = characters["devil_fruit"].fillna("Unknown")
```

```python
devil_fruits["type"].unique()
```

```python
list_type = []
for fruit in characters["devil_fruit"]:
    print(fruit)
    
    if(fruit=="Unknown"):
        list_type.append("Unknown")
    else:
        list_type.append(findtype(fruit))
print(list_type)
```

```python
list_type_df = pd.DataFrame(list_type.copy())
list_type_df = list_type_df.rename(columns={0:"devil_fruit_type"})
```

```python
list_type_df.head()
```

```python
characters = characters.join(list_type_df)
```

```python
characters.to_csv("characters_current.csv")
```
